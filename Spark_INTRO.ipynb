**Step 11 — Create RDD with 4 Partitions**

```python
data = list(range(1, 21))

rdd = sc.parallelize(data, 4)

print("Data:", rdd.collect())
print("Number of Partitions:", rdd.getNumPartitions())
```

**Step 12 — See What is Inside Each Partition**

```python
partitions = rdd.glom().collect()
#glom() is an RDD transformation that takes all the elements inside each partition and converts them into a list.
# en----> value with index
for i, partition in enumerate(partitions):
    print("Partition", i + 1, ":", partition)

    # collect() is an RDD action. Its job is: Bring all the elements of an RDD from the worker nodes back to the driver program.
```

**Step 13 — RDD Transformation: map()**

```python
data = [1, 2, 3, 4, 5]

rdd = sc.parallelize(data)

mapped_rdd = rdd.map(lambda x: x * 2)

print(mapped_rdd.collect())
```

**Step 14 — RDD Transformation: filter()**

```python
data = [1, 2, 3, 4, 5, 6, 7, 8]

rdd = sc.parallelize(data)

even_rdd = rdd.filter(lambda x: x % 2 == 0)

print(even_rdd.collect())
```

**Step 15 — Combine map() and filter()**

```python
data = [1, 2, 3, 4, 5, 6]

rdd = sc.parallelize(data)

result = rdd .map(lambda x: x * 2).filter(lambda x: x > 5)

print(result.collect())
```

**Step 16 — count() Action**

```python
data = [10, 20, 30, 40, 50]

rdd = sc.parallelize(data)

print("Number of elements:", rdd.count())
```

**Step 17 — first() Action**

```python
print("First element:", rdd.first())
```

**Step 18 — take() Action**

```python
print("First 3 elements:", rdd.take(3))
```

**Step 19 — reduce() Action**

```python
data = [1, 2, 3, 4, 5]

rdd = sc.parallelize(data)

total = rdd.reduce(lambda x, y: x + y)

print("Total:", total)
```

Step 20 — Transformations vs Actions
| Transformations | Actions     |
| --------------- | ----------- |
| `map()`         | `collect()` |
| `filter()`      | `count()`   |
| `flatMap()`     | `first()`   |
| `distinct()`    | `take()`    |
| `reduceByKey()` | `reduce()`  | **bold text**


**Step 21 — Demonstrate Lazy Evaluation**

```python
data = [1, 2, 3, 4, 5]

rdd = sc.parallelize(data)

step1 = rdd.map(lambda x: x * 2)

step2 = step1.filter(lambda x: x > 5)

print("Transformations have been defined.")

print("Now calling an action...")

result = step2.collect()

print("Result:", result)
```

# ** Word Count Using Spark**

**Create Text RDD**

```python
text = [
    "Big Data Big Data Spark"
]

text_rdd = sc.parallelize(text)

print(text_rdd.collect())
```

**Split the Sentence**

```python
words = text_rdd.flatMap( lambda line: line.split())

print(words.collect())
```

**Create Key-Value Pairs**

```python
word_pairs = words.map(lambda word: (word, 1))

print(word_pairs.collect())
```

**Count Words**

```python
word_count = word_pairs.reduceByKey(lambda x, y: x + y)

print(word_count.collect())
```

**Demonstrate DataFram**e

```python
student_data = [
    (1, "Ram", "CSE", 85),
    (2, "Sita", "CSE", 90),
    (3, "Hari", "AI", 70),
    (4, "Gita", "AI", 95)
]

columns = ["ID","Name","Department","Marks"]

df = spark.createDataFrame(student_data,columns)

df.show()
```

**Display DataFrame Schema**

```python
df.printSchema()
```

**Select Columns**

```python
df.select("Name","Marks").show()
```

**Filter DataFrame**

```python
df.filter(
    df.Marks > 80
).show()
```

**Group By Department**

```python
df.groupBy("Department").count().show()
```

**Calculate Average Marks**

```python
from pyspark.sql.functions import avg

df.select(
    avg("Marks").alias("Average Marks")
).show()
```

**Department-Wise Average**

```python
df.groupBy( "Department").agg(avg("Marks").alias("Average Marks")).show()
```

**Spark SQL**

```python
df.createOrReplaceTempView("students")
```

```python
result = spark.sql(""" SELECT * FROM students""")

result.show()
```

```python
spark.sql("""SELECT Name, Marks FROM students WHERE Marks > 80""").show()
```
