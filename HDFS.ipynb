```python
import os
import shutil
import time
```

```python
HDFS = "mini_hdfs"
```

```python
def step(title):
    print("\n" + "=" * 50)
    print(f"  {title}")
    print("=" * 50)


def pause():
    time.sleep(1)
```

```python
# STEP 1: Create HDFS directory
step("STEP 1: CREATE HDFS DIRECTORY")

print("$ hdfs dfs -mkdir /demo")
os.makedirs(HDFS, exist_ok=True)

print("✓ HDFS directory '/demo' created")
pause()
```

```python
# STEP 2: Create a local file
step("STEP 2: CREATE LOCAL FILE")

print("$ echo 'Hello HDFS!' > input.txt")

with open("input.txt", "w") as f:
    f.write("Hello HDFS!\n")
    f.write("This file is stored in our simulated HDFS.\n")
    f.write("Hadoop makes big data processing possible.")

print(" Local file 'input.txt' created")
pause()
```

```python
# STEP 3: Put file into HDFS
# --------------------------------------------------
step("STEP 3: UPLOAD FILE TO HDFS")

print("$ hdfs dfs -put input.txt /demo/")

shutil.copy("input.txt", HDFS)

print("input.txt uploaded to HDFS")
pause()
```

```python
# STEP 4: List files
step("STEP 4: LIST HDFS FILES")

print("$ hdfs dfs -ls /demo")
print()

files = os.listdir(HDFS)

for file in files:
    size = os.path.getsize(os.path.join(HDFS, file))
    print(f"-rw-r--r--  {size:5} bytes  {file}")

pause()
```

```python
# STEP 5: Read file
step("STEP 5: READ FILE FROM HDFS")

print("$ hdfs dfs -cat /demo/input.txt")
print()

with open(os.path.join(HDFS, "input.txt"), "r") as f:
    print(f.read())

pause()
```

```python
# STEP 6: Download file
step("STEP 6: DOWNLOAD FILE FROM HDFS")

print("$ hdfs dfs -get /demo/input.txt output.txt")

shutil.copy(
    os.path.join(HDFS, "input.txt"),
    "output.txt"
)

print("✓ File downloaded as 'output.txt'")
pause()
```

```python
# STEP 7: Delete file
# --------------------------------------------------
step("STEP 7: DELETE FILE FROM HDFS")

print("$ hdfs dfs -rm /demo/input.txt")

os.remove(os.path.join(HDFS, "input.txt"))

print(" input.txt deleted from HDFS")
pause()
```

```python
# STEP 8: Show final state

step("STEP 8: FINAL HDFS STATUS")

print("$ hdfs dfs -ls /demo")
print()

if os.listdir(HDFS):
    for file in os.listdir(HDFS):
        print(file)
else:
    print("HDFS directory is empty.")

print("\n HDFS DEMO COMPLETED!")
```
