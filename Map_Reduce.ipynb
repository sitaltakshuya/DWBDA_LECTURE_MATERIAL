```python
# INPUT DATA
documents = [
    "big data is powerful",
    "big data is useful",
    "hadoop processes big data",
    "hadoop is powerful"
]
#Decorative
print("=" * 70)
print("STEP 1: INPUT DATA")
print("=" * 70)
# enum--->returns--->data-with-->index
for i, document in enumerate(documents, start=1):
    print(f"Document {i}: {document}")
```

```python
# MAPPER

print("\n" + "=" * 70)
print("STEP 2: MAPPER")
print("=" * 70)

mapped_data = []

for document_id, document in enumerate(documents, start=1):

    words = document.split()

    print(f"\nMapper processing Document {document_id}:")
    
    for word in words:
        pair = (word, 1)
        mapped_data.append(pair)

        print(f"  {word}  -->  ({word}, 1)")
```

```python
# KEY VALUE PAIR

print("\n" + "=" * 70)
print("STEP 3: INTERMEDIATE KEY-VALUE PAIRS")
print("=" * 70)

for key, value in mapped_data:
    print(f"({key}, {value})")
```

```python
print("\n" + "=" * 70)
print("STEP 4: SHUFFLE")
print("=" * 70)

shuffle_data = {}

for key, value in mapped_data:

    if key not in shuffle_data:
        shuffle_data[key] = []

    shuffle_data[key].append(value)


print("\nAfter grouping same keys:")

for key, values in shuffle_data.items():
    print(f"{key} --> {values}")
```

```python
# SORT
print("\n" + "=" * 70)
print("STEP 5: SORT")
print("=" * 70)

sorted_data = dict(sorted(shuffle_data.items()))

print("\nKeys are sorted alphabetically:")

for key, values in sorted_data.items():
    print(f"{key} --> {values}")
```

```python
# REDUCER
print("\n" + "=" * 70)
print("STEP 6: REDUCER")
print("=" * 70)

reduced_data = {}

for key, values in sorted_data.items():

    total = sum(values)

    reduced_data[key] = total

    print(f"\nReducer processing:")
    print(f"  Key    = {key}")
    print(f"  Values = {values}")
    print(f"  Calculation = {' + '.join(map(str, values))}")
    print(f"  Result = {total}")
```

```python
# FINAL OUTPUT
print("\n" + "=" * 70)
print("STEP 7: FINAL OUTPUT")
print("=" * 70)

print("\nWord\t\tFrequency")
print("-" * 30)

for key, value in reduced_data.items():
    print(f"{key:<15}{value}")
```
